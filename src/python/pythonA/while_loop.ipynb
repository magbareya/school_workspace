# حلقة `while`

## ما هي الحلقة؟

**الحلقة** هي تنفيذ متكرر لعمليات معيّنة. أحيانًا نعرف مسبقًا كم مرة يجب أن تتكرّر الحلقة، وأحيانًا لا.

أمثلة:
- عدد التكرار معروف مسبقًا: عدد الصفوف في المدرسة في سنة معينة.
- عدد التكرار غير معروف مسبقًا: عدد الطلاب الذين يصلون للصف كل يوم.

في حلقة `while` يتم فحص **تعبير منطقي**: إذا تحقق الشرط تُنفَّذ الأوامر ثم نعود لفحص الشرط من جديد، وإذا لم يتحقق الشرط تتوقف الحلقة.

## الفرق بين حلقة `for` وحلقة `while`

- حلقة `for`: تُنفِّذ عدد مرات **معروف مسبقًا**.
- حلقة `while`: نستخدمها عندما لا يكون عدد مرات التكرار معروفًا مسبقًا (تنفيذ متكرر مشروط). الحلقة تتنفذ طالما الشرط يتحقق، وفي اللحظة التي لا يتحقق فيها شرط التوقف الحلقة تتوقف.

يمكن أيضًا استعمال حلقة `while` في الحالات التي نعرف فيها عدد مرات التكرار مسبقًا، تمامًا كما في حلقة `for`. على سبيل المثال، الطباعة 5 مرات:

**أ** — حلقة `while`:
```python
i = 0
while i < 5:
    print(i)
    i = i + 1
```
المتغير `i` يُستخدم كعداد للحلقة. القيمة الابتدائية = 0، شرط التوقف `i < 5`. يجب أن نحرص على تكبير `i` بـ 1.

**ب** — حلقة `for`:
```python
for i in range(0, 5, 1):
    print(i)
```
في حلقة `for` يتم تكبير `i` بـ 1 تلقائيًا.

## شرط التوقف

### شرط الحلقة

العملية تتنفذ طالما شرط الحلقة يتحقق. في المثال التالي شرط الحلقة هو `n < 10`، والمتغير `n` هو **متغير الحلقة** الذي وظيفته الإشارة إلى إيقاف تكرار الحلقة.

شغّلوا قطعة الكود وتحققوا من الناتج:

````python
n = 0
while n < 10:
    print("hi")
    n = n + 1
print("End loop")
````

## الحلقة اللانهائية

الحلقة التي تتنفذ بدون توقف تسمى **حلقة لا نهائية**.

مثال:
```python
count = 0
while count < 5:
    print(count, end=" ")
    count -= 1
```
هذه الحلقة تتنفذ بدون توقف، لأنه لا يمكننا أبدًا الوصول إلى حالة يكون فيها المتغير `count` أكبر من 5.

## دعونا نلخّص

- إذا كان شرط حلقة `while` لا يتحقق، فمقطع البرنامج لا يتنفذ.
- نستخدم حلقة `while` عندما يكون عدد مرات تنفيذ مقطع البرنامج غير معروف مسبقًا.
- يمكننا أيضًا استخدام حلقة `while` عندما يكون عدد مرات تنفيذ الحلقة معروفًا مسبقًا، كما في حلقة `for`.
- يمكننا كتابة تعبير مركب الذي يشمل عدة شروط بواسطة علاقات منطقية.
- يجب تجنب الحالة التي لا تنتهي فيها الحلقة، أي حلقة لانهائية.
